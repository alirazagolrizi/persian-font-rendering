# Persian Font and Text Rendering Setup | تنظیم فونت و نمایش متن فارسی

Configure Matplotlib fonts and optional Persian text shaping helpers for downstream plots.

Open the repository README for prerequisites, input assets, limitations, and run instructions. | برای پیش‌نیازها، ورودی‌ها و راهنمای اجرا README مخزن را ببینید.

```python
# اجرا کن قبل از رسم نمودارها
import matplotlib.pyplot as plt
import matplotlib
from pathlib import Path

# اگر می‌خواهی از فونت خاصی استفاده کنی، مسیر ttf را اینجا قرار بده
preferred_fonts = ["Vazir.ttf", "Tahoma.ttf", "DejaVuSans.ttf"]

font_path = None
for f in preferred_fonts:
    p = Path(f)
    if p.exists():
        font_path = str(p.resolve())
        break

if font_path:
    matplotlib.font_manager.fontManager.addfont(font_path)
    matplotlib.rcParams['font.family'] = Path(font_path).stem
else:
    # fallback
    matplotlib.rcParams['font.family'] = 'DejaVu Sans'

# برای راست‌چین و شکل‌دهی بهتر متن فارسی (اختیاری)
try:
    import arabic_reshaper
    from bidi.algorithm import get_display
    def persian_text(s):
        return get_display(arabic_reshaper.reshape(s))
    reshaper_available = True
except Exception:
    def persian_text(s): return s
    reshaper_available = False

print("فونت انتخاب‌شده:", matplotlib.rcParams['font.family'])
print("arabic_reshaper available:", reshaper_available)
```
